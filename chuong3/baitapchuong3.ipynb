```python
from colletions import deque
from typing import Dict, List, Optional

def truy_vet(vpred: Dict[int, Optional[int]], bat_dau: int, ket_thuc: int) -> List[Int]:
    if ket_thuc not in vpred and ket_thuc != bat_dau:
        return []
    
    duong_di = []
    dinh_hien_tai = ket_thuc
    while dinh_hien_tai is not None:
        duong_di.append(dinh_hien_tai)
        dinh_hien_tai = vpred[dinh_hien_tai]
    
    return duong_di[::-1]
```

```python
def tim_dfs(do_thi: Dict[int, Optional[int]], bat_dau: int, ket_thuc: int) -> List[Int]:
    if bat_dau not in do_thi and ket_thuc not in do_thi:
        return []
    if bat_dau == ket_thuc:
        return [bat_dau]
    
    ngan_xep = [bat_dau]
    da_tham = {bat_dau}
    vpred = {bat_dau: None}
    
    tim_thay = False
    while ngan_xep:
        hien_tai = ngan_xep.pop()
        if hien_tai == ket_thuc:
            tim_thay =True
            break
        for ke in do_thi.get(hien_tai, []):
            if ke not in da_tham:
                da_tham.add(ke)
                vpred[ke] = hien_tai
                ngan_xep.append(ke)
    if not tim_thay:
        return []
    return truy_vet(vpred, bat_dau, ket_thuc)
```

```python
def tim_bfs(do_thi: Dict[int, Optional[int]], bat_dau: int, ket_thuc: int) -> List[Int]:
    if bat_dau not in do_thi and ket_thuc not in do_thi:
        return []
    if bat_dau == ket_thuc:
        return[bat_dau]
    
    hang_doi = deque([bat_dau])
    da_tham = {bat_dau}
    vpred = {bat_dau: None}
    
    while hang_doi:
        hien_tai = hang_doi.popleft()
        if hien_tai == ket_thuc:
            break
        for ke in do_thi.get(hien_tai,[]):
            if ke not in da_tham:
                da_tham.add(ke)
                vpred[ke] = hien_tai
                hang_doi.append(ke)
    return truy_vet(vpred, bat_dau, ket_thuc)
```

```python
do_thi_mau = {
    1: [2, 3],
    2: [1, 3, 4],
    3: [1, 2],
    4: [2, 6],
    5: [6],
    6: [4, 5],
    7: []  #Dinh co lap de kiem tra khong co duong di
}

diem_dau = 1
diem_cuoi = 5

duong_bfs = tim_bfs(do_thi_mau, diem_dau, diem_cuoi)
duong_dfs = tim_dfs(do_thi_mau, diem_dau, diem_cuoi)

print(f"Ket qua tim duong tu {diem_dau} den {diem_cuoi}")
print(f"Duong di BFS (Toi uu so canh) : {' -> '.join(map(str, duong_bfs)) if duong_bfs else 'Khong co duong'}")
print(f"Duong di DFS                  : {' -> '.join(map(str, duong_dfs)) if duong_dfs else 'Khong co duong'}")

#Test voi dinh co lap
duong_khong_toi = tim_bfs(do_thi_mau, 1, 7)
print(f"Duong di den dinh co lap 7    : {duong_khong_toi}")
```
