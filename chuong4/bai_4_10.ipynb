```python
from typing import List
```

```python
def sap_xep_cho_ngoi(n: int) -> List[List[int]]:
    if n < 3 or n % 2 == 0:
        return []
    m = n - 1
    ket_qua = []
    for k in range(m // 2):
        thu_tu = [0]
        for i in range(m):
            buoc = (i + 1) // 2 if i % 2 == 1 else -(i // 2)
            thu_tu.append((k + buoc) % m + 1)
        ket_qua.append(thu_tu)
    return ket_qua

def kiem_tra(cach_xep: List[List[int]]) -> bool:
    da_ngoi_canh = set()
    for vong in cach_xep:
        for i in range(len(vong)):
            cap = frozenset((vong[i], vong[(i + 1) % len(vong)]))
            if cap in da_ngoi_canh:
                return False
            da_ngoi_canh.add(cap)
    return True
```

```python
for so_nguoi in [5, 9, 11]:
    cach_xep = sap_xep_cho_ngoi(so_nguoi)
    print(f"{so_nguoi} nguoi ngoi ban tron -> {len(cach_xep)} ngay khong ai ngoi canh nguoi cu:")
    for ngay, vong in enumerate(cach_xep, 1):
        print(f"  Ngay {ngay}:", " - ".join(str(x + 1) for x in vong))
    print("  Hop le:", kiem_tra(cach_xep))
    print()
```
