```python
from collections import deque
from typing import Dict, List
```

```python
def lien_thong(do_thi: Dict[int, List[int]]) -> bool:
    co_canh = [u for u in do_thi if do_thi[u]]
    if not co_canh:
        return True
    da_tham = {co_canh[0]}
    hang_doi = deque([co_canh[0]])
    while hang_doi:
        u = hang_doi.popleft()
        for v in do_thi[u]:
            if v not in da_tham:
                da_tham.add(v)
                hang_doi.append(v)
    return all(u in da_tham for u in co_canh)

def tim_chu_trinh_euler(do_thi: Dict[int, List[int]]) -> List[int]:
    if not lien_thong(do_thi):
        return []
    if any(len(do_thi[u]) % 2 != 0 for u in do_thi):
        return []
    co_canh = [u for u in do_thi if do_thi[u]]
    if not co_canh:
        return []
    ke = {u: list(do_thi[u]) for u in do_thi}
    ngan_xep = [co_canh[0]]
    chu_trinh = []
    while ngan_xep:
        u = ngan_xep[-1]
        if ke[u]:
            v = ke[u].pop()
            ke[v].remove(u)
            ngan_xep.append(v)
        else:
            chu_trinh.append(ngan_xep.pop())
    return chu_trinh[::-1]
```

```python
g_euler = {
    1: [2, 3],
    2: [1, 3, 4, 5],
    3: [1, 2, 4, 5],
    4: [2, 3, 5, 6],
    5: [2, 3, 4, 6],
    6: [4, 5]
}
g_khong_euler = {
    1: [2, 3],
    2: [1, 3, 4],
    3: [1, 2],
    4: [2]
}

ct = tim_chu_trinh_euler(g_euler)
print("Chu trinh Euler:", " -> ".join(map(str, ct)) if ct else "Khong co")
ct2 = tim_chu_trinh_euler(g_khong_euler)
print("Do thi 2       :", " -> ".join(map(str, ct2)) if ct2 else "Khong co chu trinh Euler")
```
