```python
from typing import List, Optional
```

```python
BUOC_MA = [(2, 1), (1, 2), (-1, 2), (-2, 1), (-2, -1), (-1, -2), (1, -2), (2, -1)]

def dem_nuoc_di(ban_co: List[List[int]], x: int, y: int, n: int) -> int:
    dem = 0
    for dx, dy in BUOC_MA:
        nx, ny = x + dx, y + dy
        if 0 <= nx < n and 0 <= ny < n and ban_co[nx][ny] == 0:
            dem += 1
    return dem

def ma_di_tuan(n: int, x0: int = 0, y0: int = 0) -> Optional[List[List[int]]]:
    ban_co = [[0] * n for _ in range(n)]
    ban_co[x0][y0] = 1

    def quay_lui(x: int, y: int, buoc: int) -> bool:
        if buoc == n * n:
            return True
        ung_vien = []
        for dx, dy in BUOC_MA:
            nx, ny = x + dx, y + dy
            if 0 <= nx < n and 0 <= ny < n and ban_co[nx][ny] == 0:
                ung_vien.append((dem_nuoc_di(ban_co, nx, ny, n), nx, ny))
        ung_vien.sort()
        for _, nx, ny in ung_vien:
            ban_co[nx][ny] = buoc + 1
            if quay_lui(nx, ny, buoc + 1):
                return True
            ban_co[nx][ny] = 0
        return False

    return ban_co if quay_lui(x0, y0, 1) else None

def in_ban_co(ban_co: Optional[List[List[int]]]) -> None:
    if ban_co is None:
        print("Khong co loi giai")
        return
    for hang in ban_co:
        print(" ".join(f"{o:3d}" for o in hang))
```

```python
print("Ban co 5x5:")
in_ban_co(ma_di_tuan(5))
print()
print("Ban co 8x8:")
in_ban_co(ma_di_tuan(8))
print()
print("Ban co 4x4:")
in_ban_co(ma_di_tuan(4))
```
