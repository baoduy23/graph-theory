```python
from itertools import permutations
from typing import List, Tuple
```

```python
INF = float("inf")

def tsp_vet_can(chi_phi: List[List[float]]) -> Tuple[float, List[int]]:
    n = len(chi_phi)
    tot_nhat, duong_tot = INF, []
    for hv in permutations(range(1, n)):
        duong = [0, *hv, 0]
        tong = sum(chi_phi[duong[i]][duong[i + 1]] for i in range(n))
        if tong < tot_nhat:
            tot_nhat, duong_tot = tong, duong
    return tot_nhat, duong_tot

def tsp_quy_hoach_dong(chi_phi: List[List[float]]) -> Tuple[float, List[int]]:
    n = len(chi_phi)
    dp = [[INF] * n for _ in range(1 << n)]
    truoc = [[-1] * n for _ in range(1 << n)]
    dp[1][0] = 0
    for mask in range(1 << n):
        for u in range(n):
            if dp[mask][u] == INF or not (mask >> u) & 1:
                continue
            for v in range(n):
                if (mask >> v) & 1:
                    continue
                moi = mask | (1 << v)
                gia = dp[mask][u] + chi_phi[u][v]
                if gia < dp[moi][v]:
                    dp[moi][v] = gia
                    truoc[moi][v] = u
    full = (1 << n) - 1
    tot_nhat, cuoi = INF, -1
    for u in range(1, n):
        gia = dp[full][u] + chi_phi[u][0]
        if gia < tot_nhat:
            tot_nhat, cuoi = gia, u
    duong = [0]
    mask, u = full, cuoi
    while u != 0:
        duong.append(u)
        mask, u = mask ^ (1 << u), truoc[mask][u]
    duong.append(0)
    return tot_nhat, duong[::-1]
```

```python
ten_tp = ["A", "B", "C", "D", "E"]
chi_phi = [
    [0, 3, 4, 2, 7],
    [3, 0, 4, 6, 3],
    [4, 4, 0, 5, 8],
    [2, 6, 5, 0, 6],
    [7, 3, 8, 6, 0]
]

gia1, duong1 = tsp_vet_can(chi_phi)
gia2, duong2 = tsp_quy_hoach_dong(chi_phi)
print("Vet can       :", " -> ".join(ten_tp[i] for i in duong1), "| Chi phi =", gia1)
print("Quy hoach dong:", " -> ".join(ten_tp[i] for i in duong2), "| Chi phi =", gia2)
```
