```python
from typing import Dict, List
```

```python
def tim_chu_trinh_hamilton(do_thi: Dict[int, List[int]]) -> List[int]:
    dinh = list(do_thi.keys())
    n = len(dinh)
    if n == 0:
        return []
    bat_dau = dinh[0]
    duong = [bat_dau]
    da_tham = {bat_dau}

    def quay_lui(u: int) -> bool:
        if len(duong) == n:
            return bat_dau in do_thi[u]
        for v in do_thi[u]:
            if v not in da_tham:
                da_tham.add(v)
                duong.append(v)
                if quay_lui(v):
                    return True
                duong.pop()
                da_tham.remove(v)
        return False

    if quay_lui(bat_dau):
        return duong + [bat_dau]
    return []
```

```python
g_hamilton = {
    1: [2, 4, 5],
    2: [1, 3, 5],
    3: [2, 4, 5],
    4: [1, 3, 5],
    5: [1, 2, 3, 4]
}
g_khong_hamilton = {
    1: [2, 3, 4],
    2: [1],
    3: [1],
    4: [1]
}

ct = tim_chu_trinh_hamilton(g_hamilton)
print("Chu trinh Hamilton:", " -> ".join(map(str, ct)) if ct else "Khong co")
ct2 = tim_chu_trinh_hamilton(g_khong_hamilton)
print("Do thi 2          :", " -> ".join(map(str, ct2)) if ct2 else "Khong co chu trinh Hamilton")
```
