```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
def ve(G, ten, pos=None):
    if pos is None:
        pos = nx.circular_layout(G)
    plt.figure(figsize=(4, 4))
    nx.draw(G, pos, with_labels=True, node_color="skyblue", node_size=500)
    plt.title(f"{ten} - {G.number_of_nodes()} dinh, {G.number_of_edges()} canh")
    plt.show()
```

```python
n = 6
k = 3
x = 3
y = 4
```

```python
G = nx.complete_graph(n)
ve(G, f"K{n}")
```

```python
G = nx.cycle_graph(n)
ve(G, f"C{n}")
```

```python
G = nx.wheel_graph(n)
pos = nx.circular_layout(range(1, n))
pos[0] = (0, 0)
ve(G, f"W{n}", pos)
```

```python
G = nx.complete_bipartite_graph(x, y)
ve(G, f"K{x},{y}", nx.bipartite_layout(G, range(x)))
```

```python
if n * k % 2 == 0 and k < n:
    G = nx.random_regular_graph(k, n)
    ve(G, f"{k}-regular")
else:
    print("Khong ton tai")
```
